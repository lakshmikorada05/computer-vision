```python
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

r_min = 100
r_max = 200

img = cv2.imread("/content/images.jpg", 0)

# Check if the image was loaded successfully
if img is None:
    print("Error: Could not load image. Please check the file path.")
else:
    slicing_with_bg = img.copy()
    slicing_without_bg = np.zeros(img.shape, dtype=np.uint8)

    for i in range(img.shape[0]):
        for j in range(img.shape[1]):

            pixel_value = img[i, j]

            if r_min <= pixel_value <= r_max:
                slicing_with_bg[i, j] = 255
                slicing_without_bg[i, j] = 255

            else:
                slicing_without_bg[i, j] = 0

    plt.figure(figsize=(12, 4))

    plt.subplot(1, 3, 1)
    plt.imshow(img, cmap='gray')
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(slicing_with_bg, cmap='gray')
    plt.title("Slicing With Background")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(slicing_without_bg, cmap='gray')
    plt.title("Slicing Without Background")
    plt.axis("off")

    plt.show()
```
