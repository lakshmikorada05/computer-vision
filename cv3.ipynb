```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
#function to display images
def show_comparison(original,blur3,blur5,blur7,blur11):
  plt.figure(figsize=(15,10))
  #original image
  plt.subplot(2,3,1)
  plt.imshow(original,cmap='gray')
  plt.title('Original  grayscale Image')
  plt.axis('off')
  #3x3 kernal
  plt.subplot(2,3,2)
  plt.imshow(blur3,cmap='gray')
  plt.title(' average filter 3x3 Kernel')
  plt.axis('off')
  #5x5 kernal
  plt.subplot(2,3,3)
  plt.imshow(blur5,cmap='gray')
  plt.title(' average filter 5x5 Kernel')
  plt.axis('off')
  #7x7 kernal
  plt.subplot(2,3,4)
  plt.imshow(blur7,cmap='gray')
  plt.title(' average filter 7x7 Kernel')
  plt.axis('off')
  #11x11 kernal
  plt.subplot(2,3,5)
  plt.imshow(blur11,cmap='gray')
  plt.title(' average filter 11x11 Kernel')
  plt.axis('off')
  plt.tight_layout()
  plt.show()
```

```python
#1. read the IMAGE in gary scale
img=cv2.imread('/content/nlm_denoised.jpg',0)
if img is None:
  print('could not open the image')
else:
  #2.define avg kernal of diff size(3x3,5x5,7x7 11x11)
  kernel_3x3=np.ones((3,3),np.float32)/9
  kernel_5x5=np.ones((5,5),np.float32)/25
  kernel_7x7=np.ones((7,7),np.float32)/49
  kernal_11x11=np.ones((11,11),np.float32)/121
  #3.apply each kernel to image using 2D convultion operation
  blurred_3x3=cv2.filter2D(img,-1,kernel_3x3)
  blurred_5x5=cv2.filter2D(img,-1,kernel_5x5)
  blurred_7x7=cv2.filter2D(img,-1,kernel_7x7)
  blurred_11x11=cv2.filter2D(img,-1,kernal_11x11)
  #4.DISPLAYthe original images
  show_comparison(img,blurred_3x3,blurred_5x5,blurred_7x7,blurred_11x11)
```
